```python
from scipy import linalg

mat = np.array([[2, 4], [1, 3]])
print(linalg.det(mat))
```

```python
eig_val, eig_vec = linalg.eig(mat)
print(eig_val)
print(eig_vec)
```

```python
from scipy import stats

marks = df1['marks']
print(marks.mean())
print(marks.median())
print(stats.mode(marks))
print(marks.var())
print(marks.std())
```

```python
reshaped = df1.values.reshape(3,1,3)
print(reshaped)
```

```python
filtered = df1[df1['marks'] > 55]
print(filtered)
```

```python
extra = pd.DataFrame({'id':[4,5], 'name':['D','E'], 'marks':[80,90]})
merged = pd.concat([df1, extra])
print(merged)
```

```python
df_missing = df1.copy()
df_missing.loc[1,'marks'] = np.nan
print(df_missing.fillna(df_missing['marks'].mean()))
```
